# Creating Atoms and Accessing Properties

This tutorial explains how to create atoms and access their relevant properties. These atoms are created in MDMC using a module called 'periodictable' and all of the properties associated with the atoms are now accessed via the atom objects. This module allows for various isotopes to be specified as well as the natural abundances. To conclude this tutorial there is a short example of creating a water molecule with deuterium instead of hydrogen.

```python
# imports used for this tutorial
from MDMC.MD.structures import Atom
from MDMC.trajectory_analysis.observables.fqt import calc_incoherent_scatt_length
from MDMC.MD import Molecule, BondAngle
from MDMC.MD import Bond
from MDMC.MD import HarmonicPotential
```

Now that the Atom class has been imported, we can create an atom. This can be done to create an atom which represents the natural abundance of the element, or a specific isotope.

```python
# Creating an oxygen atom using MDMC Atom class

# Standard chemical symbol will give natural abundance
oxygen_na = Atom('O')

# Chemical symbol followed by the mass number will give the specific isotope
oxygen_16 = Atom('O[16]')

# Example for specifying different isotope
oxygen_17 = Atom('O[17]')
```

Now that the atoms are created we can start accessing their associated properties .

```python
# Accessing general properties
mass                  = oxygen_na.mass
atomic_number         = oxygen_na.element.number

# Accessing scattering specific properties, like scattering lengths and cross sections
coh_cross_section     = oxygen_na.element.neutron.coherent
incoh_cross_section   = oxygen_na.element.neutron.incoherent
coh_scatt_length      = oxygen_na.element.neutron.b_c

# Many incoherent scattering lengths are stored as 0, but with non-zero cross sections.
# The lengths can be approximated with the following function, using oxygen 'O' as an example again.
incoh_scatt_length    = calc_incoherent_scatt_length('O')

# To see all attributes, use dir(), e.g dir(oxygen_na.element)
```

Now lets create a heavy water molecule, by using deuterium instead of hydrogen.

```python
# Conveniently, deuterium can be specified directly through this artificial chemical symbol 'D' 
# It can also be specified using 'H[2]' as expected.

D1 = Atom('D')
D2 = Atom('D', position=[0., 1.63298, 0.])
O = Atom('O', position=[0., 0.81649, 0.57736])

HOH_angle = BondAngle(D1, O, D2)
HO_bond = Bond(D1, O, function=HarmonicPotential(1., 100., interaction_type='bond'))

water_mol = Molecule(position=[2., 1.5, 1.], atoms=[D1, D2,O], interactions=[HO_bond])

```
