# Dumping H5MD trajectory files
H5MD files are created in the refinement step of a simulation. To get to this step of MDMC `Universe`, `Simulation` and `Control` class's needs to be created.

In order for this to be created the following needs to be imported:

```python
from MDMC.control import Control
from MDMC.MD import Simulation, Universe
```

The `Simulation`, and `Universe`, needs to be built as explained further [here](building-a-universe-checkpoint.ipynb)

```python
universe = Universe(dimensions=38.4441)

simulation = Simulation(universe,
                        engine="lammps",
                        time_step=10.18893,
                        temperature=120.,
                        traj_step=15)
```

A `Control` variable then needs to be declared but `Control` needs a dataset and universe parameters.

```python
exp_datasets = [{'file_name':'data/Well_s_q_omega_Ar_data.xml',
                 'type':'SQw',
                 'reader':'xml_SQw',
                 'weight':1.,
                 'auto_scale':True,
                 'resolution':800}]

fit_parameters = universe.parameters
```

At the point of declaring `Control` to make the refinement dump the H5MD files `Dump` needs to be set to ether 1 to dump every trajectory or -1 to dump the best trajectory.

```python
control_best = Control(simulation=simulation,
                  exp_datasets=exp_datasets,
                  fit_parameters=fit_parameters,
                  MD_steps=4000,
                  h5md_dump=-1)
```

```python
control_all = Control(simulation=simulation,
                  exp_datasets=exp_datasets,
                  fit_parameters=fit_parameters,
                  MD_steps=4000,
                  h5md_dump=1)
```

Finally, `refine` should be run to write the trajectory.

Running `refine` with only `n_steps` declared will result in trajectories being written, however, there are some further settings you can declare to change how and where the trajectories are written:

* `h5md_file_name`, will change the name of the file, default is trajectory.h5
* `h5md_file_loc`, will change where the file is written
* `h5md_timestamp`, will change if there is a timestamp within the file name, default is `True`

```python
control_best.refine(n_steps=5, h5md_file_name = "Dump-Best.h5", h5md_file_loc = "/", h5md_timestamp = False)
```

This code above will result in a file called `Dump-Best.h5` being created in the current file directory

Note
----
The file extension `.h5` will not be by added by default but it is suggested for file organisation and easier identification.

Warning
-------

If dumping all trajectoryies it is recommended that `h5md_timestamp` is set to `True` as, if only one file is defined, it will continually overwrite the last trajectory meaning only the last trajectory will be stored.


```python
control_all.refine(n_steps=5, h5md_file_name = "Dump-All.h5", h5md_file_loc = "/", h5md_timestamp = True)
```
